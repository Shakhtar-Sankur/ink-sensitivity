# Scroll ink sensitivity, stage 2: real ink, transplanted

Stage 1 planted letters drawn as a pure brightness shift (the measured mean profile of PHerc. 139 ink). The official
model barely responded, even at 4× strength, while it responds strongly to the real text next to them: it keys on what
ink looks like, not on brightness alone. So stage 2 moves **real ink**.

On **PHerc. 139** the published 2.4 µm map labels where the ink is. We take 2.2 mm windows of that real text and write
the 9 µm voxel columns under the labelled ink, texture and all, into the papyrus of the target scroll through the ink's own
stroke shapes, matched layer by layer to the target's brightness and spread. The control does exactly the same with
**bare PHerc. 139 papyrus** through the same stroke shapes, so ink and control differ only in the ink. A fifth of the
windows are left untouched.

- **PHerc. 139** (calibration: also scores the same real text in place, untouched)
- **PHerc. 800**: eligible for First Letters and the Grand Prize; five published segments

A window counts as recovered when it beats the 95th percentile of the bare-papyrus controls. The AUC is the chance that a
real-ink window outscores a bare one (0.5 = blind, 1 = perfect). Settings: **GPU T4 x2**, Internet **on**.
Best: **Save Version → Save & Run All**, about 1 hour.

```python
import glob
import json
import os
import shutil
import subprocess
import time

from huggingface_hub import hf_hub_download

W = "/kaggle/working"
ENV = "/kaggle/working/villa/ink-detection"  # the official environment (zarr 2, tifffile, imagecodecs)


def run(*cmd):
    r = subprocess.run(["uv", "run", "python", *cmd], cwd=ENV, capture_output=True, text=True)
    if r.returncode != 0:
        print(r.stdout[-2000:], r.stderr[-3000:])
        raise RuntimeError(f"failed: {cmd[:3]}")
    return r.stdout


print(run(f"{W}/inject.py", "signature", "--out", f"{W}/sig.json"))
VOLS = []
for seed in (0, 1, 2):
    for t in ["0139"] + [f"0800-{i}" for i in range(5)]:
        d = f"{W}/inj/{t}_s{seed}"
        os.makedirs(d, exist_ok=True)
        print(run(f"{W}/inject.py", "make", "--sig", f"{W}/sig.json", "--target", t, "--out", d, "--seed", str(seed),
                  "--mode", "transplant").strip())
        VOLS.append(d)
for f in ("donor_vol.npy", "donor_ref.npy"):  # only needed to build the volumes
    os.remove(f"{W}/{f}")
```

```python
# Official inference on every injected volume: two checkpoints, both depth directions
CKPTS = [hf_hub_download("scrollprize/ink_9um", f)
         for f in ("hybrid_3d2d-seed42/step-075000.pth", "hybrid_3d2d-seed43/step-020000.pth")]
DECODE = ("import glob, sys, numpy as np, tifffile; "
          "[np.save(p.rsplit('.', 1)[0] + '.npy', tifffile.imread(p)) for p in glob.glob(sys.argv[1] + '/ck*.tif*')]")
t_all = time.time()
for d in VOLS:
    t0 = time.time()
    ok = True
    for i, ck in enumerate(CKPTS):
        try:  # a volume normally takes 2-3 min; skip one that hangs instead of blocking the run
            r = subprocess.run(["uv", "run", "python", "-m", "koine_machines.inference.infer", f"{d}/volume.zarr", ck,
                                f"{d}/ck{i}.tif", "--direction", "both", "--batch-size", "16"],
                               cwd=ENV, capture_output=True, text=True, timeout=900)
        except subprocess.TimeoutExpired:
            print(f"timeout on {d}, checkpoint {i}: skipped")
            ok = False
            break
        if r.returncode != 0:
            print(f"inference failed on {d}: skipped\n", r.stdout[-1500:], r.stderr[-3000:])
            ok = False
            break
    if not ok:
        continue
    subprocess.run(["uv", "run", "python", "-c", DECODE, d], cwd=ENV, check=True)
    preds = sorted(glob.glob(f"{d}/ck*.npy"))
    print(run(f"{W}/inject.py", "score", "--dir", d, "--pred", *preds).splitlines()[0], f"({time.time() - t0:.0f} s)")
    shutil.rmtree(f"{d}/volume.zarr")  # keep the saved output small
print(f"all inference: {(time.time() - t_all) / 60:.1f} min")
```

```python
done = [d for d in VOLS if os.path.exists(f"{d}/scores.json")]
print(len(done), "of", len(VOLS), "volumes scored")
print(run(f"{W}/inject.py", "aggregate", "--dirs", *[d for d in done if "/0139_" in d],
          "--label", "PHerc. 139 (calibration)", "--out", f"{W}/agg_0139.json"))
print(run(f"{W}/inject.py", "aggregate", "--dirs", *[d for d in done if "/0800-" in d],
          "--label", "PHerc. 800 (eligible, unread)", "--out", f"{W}/agg_0800.json"))
```

```python
# Picture: model output on one PHerc. 800 segment; red = real ink moved in, blue = bare papyrus moved in, grey = untouched
import numpy as np
from IPython.display import Image as Show
from IPython.display import display
from PIL import Image, ImageDraw

d = sorted(d for d in VOLS if "/0800-3_" in d and os.path.exists(f"{d}/scores.json"))[0]
meta = json.load(open(f"{d}/letters.json"))
pred = np.mean([np.load(p).astype(np.float32) for p in sorted(glob.glob(f"{d}/ck*.npy"))], axis=0) / 255.0
img = Image.fromarray((np.clip((pred - 0.25) / 0.5, 0, 1) * 255).astype(np.uint8)).convert("RGB")
dr = ImageDraw.Draw(img)
s = meta["window_px"]
colour = {"ink": (255, 60, 60), "bare": (80, 160, 255), "empty": (150, 150, 150), "real": (255, 200, 0)}
for e in meta["entries"]:
    dr.rectangle([e["x"], e["y"], e["x"] + s, e["y"] + s], outline=colour[e["kind"]], width=5)
img.thumbnail((1600, 1600))
img.save(f"{W}/pherc0800_transplant_example.png")
display(Show(filename=f"{W}/pherc0800_transplant_example.png", width=1000))
```
